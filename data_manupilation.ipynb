```python
age_goruped = pd.crosstab(df_surver['age_group'], df_surver['alcohol_consumption_change_12m'])
age_goruped.to_csv('datas/age_grouped_consum.csv')
```

```python
def change_to_short(discovery):
    if discovery in ['Google-Suche','Instagram/Social Media']:
        return 'online'
    if discovery in ['Empfehlung (Word of Mouth)', 'Sonstiges']:
        return 'heard'
    if discovery in ['Wochenmarkt/Event', 'Fachhandel-Regal' ]:
        return 'physical contact'


df_surver['psy_cat'] = df_surver['discovery_channel'].apply(change_to_short)

df_surver.head(20)
```

```python
grouped_psy_and_nps = df_surver.groupby('psy_cat')['nps_score_0_10']
grouped_mean_nps = grouped_psy_and_nps.mean().reset_index().sort_values(by='nps_score_0_10' ,ascending=False)
grouped_mean_nps= grouped_mean_nps.rename(columns={'nps_score_0_10':'mean_of_nps_score'})
grouped_mean_nps['mean_of_nps_score'] = grouped_mean_nps['mean_of_nps_score'].round(2)
grouped_mean_nps
```

```python
prefer_pre_psy_cat = pd.crosstab(df_surver['psy_cat'], df_surver['preferred_purchase_channel'])
prefer_pre_psy_cat.sort_values(by='Hofladen (vor Ort)', ascending=False).reset_index().to_csv('datas/prefer_to_find.csv')
```
